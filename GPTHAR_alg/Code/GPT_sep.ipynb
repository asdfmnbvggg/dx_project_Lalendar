```python
def load_data(data):
    train_x = None
    train_y = None
    # Load train data X from the pickle file
    # with open(data+'_train_GPT_data_X.pickle', 'rb') as f:
    #    train_x =  np.array(pickle.load(f),dtype=object)

    # Load train data Y from the pickle file
    # with open(data+'_train_GPT_data_Y.pickle', 'rb') as f:
    #    train_y = np.array(pickle.load(f),dtype=object)

    # Load the list of DataFrames from the pickle file
    with open("datasets/"+data+'_train_data.pickle', 'rb') as f:
        train_x = pickle.load(f)

    return train_x, train_y
```

```python
def sliding_window(elements, window_size, stride, wordDict):
        chunks_x = []
        chunks_y = []

        if len(elements) < window_size:

            #elements = np.insert(elements, 0, wordDict['<endoftext>'])
     

            x = elements[:-1]
            y = elements[1:]

            #print("")
            #print("INFERIEUR !!!!!!")
            #print("")
            #print(len(elements))
            #print("")
            #print(elements)
            #print("")
            #print(x)
            #print("")
            #print(y)
            #input("Press Enter to continue...")

            chunks_x.append(x)
            chunks_y.append(y)

            return chunks_x, chunks_y
        
        n_windows = int(np.ceil((len(elements) - window_size) / stride))+1
        
        for i in range(n_windows):

            

            start = i * stride
            end = start + window_size
            if end > len(elements):
                start -= (end - len(elements))
                end = len(elements)

            #x = elements[start:end][:-1]
            #y = elements[start:end][1:]
            sub_seq= elements[start:end]
            #sub_seq = np.insert(sub_seq, 0, wordDict['<endoftext>'])
            x = sub_seq[:-1]
            y = sub_seq[1:]

            chunks_x.append(x)
            chunks_y.append(y)

        return chunks_x, chunks_y
```

```python
def encode(df):
        df['merge'] = df['sensor'].astype(str) + df['value'].astype(str)

        feature_1 = df['merge'].values.astype(str)
        #activity = df['activity'].values[0]
        #activity = df['activity'].value_counts().idxmax()
        
        return feature_1
```

```python
def split_dataset_into_segments(df):

    chunks = []

    transitionIndex = df.activity.ne(df.activity.shift())

    ii = np.where(transitionIndex == True)[0]

    for i, end in enumerate(ii):
        if i > 0:
            start = ii[i - 1]

            # activitySeq = self.df[start:end]
            if(len(df[start:end])>1):
                chunks.append(df[start:end])

    # lastActivitySeq = self.df[end:]

    if(len(df[end:])>1):
        chunks.append(df[end:])

    return chunks
```

```python
dataset_name = "aruba"
```

```python
# Load data
train_x, train_y = load_data(dataset_name)
```

```python
len(train_x)
```

```python
print("Turn sensor into Words")
training_weeks = []

for df_week in train_x :
    activities_seqs_in_week = split_dataset_into_segments(df_week)

    encoded_seqs = []
    for seq in activities_seqs_in_week:
        x = encode(seq)
        #encoded_seqs.append(np.concatenate((['<Start>'], x, ['<End>']), axis = 0))
        encoded_seqs.append(np.concatenate((x, ['<SEP>']), axis = 0))

    training_weeks.append(encoded_seqs)


```

```python
print("Turn sensor words into sentences")
training_weeks_text = []
for week in training_weeks:
    #print(week)
    #s = np.concatenante(week)
    s = np.concatenate((week), axis = None)
    #print(s)
    training_weeks_text.append(" ".join(s))
```

```python
training_weeks_text
```

```python
print("Tokenize and create vocabulary")
#print(len(training_weeks_text))
tokenizer = Tokenizer(filters = '', lower = False, oov_token="<UNK>")
tokenizer.fit_on_texts(training_weeks_text)
wordDict = tokenizer.word_index
```

```python
wordDict
```

```python
wordDict['<endoftext>'] = len(wordDict)+1
```

```python
wordDict
```

```python
wordDict['<SEP>']
```

```python
print("Encode sentences")
training_weeks_encoded = tokenizer.texts_to_sequences(training_weeks_text)
```

```python
print("Create Input and Output to train the model")
input_sequences_encoded = []
output_sequences_encoded = []
```

```python
bloc_size = 1024
stride = 512

for week in training_weeks_encoded:

    x,y = sliding_window(week, bloc_size+1,stride, wordDict)
    
    print(np.array(x).shape)
    
    #if len(x) < 1024:
    #    print(x)
    #    print("")
    #    print(y)
    #    input("Press Enter to continue...")
    input_sequences_encoded = input_sequences_encoded + x
    output_sequences_encoded = output_sequences_encoded + y
```
