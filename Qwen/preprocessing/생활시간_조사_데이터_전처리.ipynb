```python
import pandas as pd

df2019 = pd.read_csv('/content/drive/MyDrive/시간대_전체_20260617_09018_데이터/2019_시간대_전체_20260617_09018.csv', encoding='cp949')

df2024 = pd.read_csv('/content/drive/MyDrive/시간대_전체_20260617_09018_데이터/2024_시간대_전체_20260617_09018.csv', encoding='cp949')
```

```python
print(f"Number of columns in df2019: {df2019.shape[1]}")
print(f"Number of columns in df2024: {df2024.shape[1]}")
```

```python
print(len(df2019))
print(len(df2024))
```

```python
# Get column names for both DataFrames
columns_2019 = set(df2019.columns)
columns_2024 = set(df2024.columns)

# Find columns unique to df2019
unique_to_2019 = list(columns_2019 - columns_2024)
print(f"Columns unique to df2019: {unique_to_2019}")

# Find columns unique to df2024
unique_to_2024 = list(columns_2024 - columns_2019)
print(f"Columns unique to df2024: {unique_to_2024}")
```

```python
# Drop columns unique to df2019 from df2019
df2019 = df2019.drop(columns=unique_to_2019, errors='ignore')
print(f"df2019 columns after dropping: {df2019.columns.tolist()}\n")

# Drop columns unique to df2024 from df2024
df2024 = df2024.drop(columns=unique_to_2024, errors='ignore')
print(f"df2024 columns after dropping: {df2024.columns.tolist()}")
```

```python
print(f"Number of columns in df2019: {df2019.shape[1]}")
print(f"Number of columns in df2024: {df2024.shape[1]}")
```

```python
df2019.columns
```

```python
# Concatenate df2019 and df2024
df_combined = pd.concat([df2019, df2024], ignore_index=True)

print("Combined DataFrame head:")
display(df_combined.head())

print(f"\nShape of the combined DataFrame: {df_combined.shape}")
```

```python
nan_counts = df_combined.isnull().sum()
nan_columns = nan_counts[nan_counts > 0]

print("Columns with NaN values and their counts in df_combined:")
display(nan_columns)
```

```python
threshold = 0.50 * len(df_combined)
columns_to_drop = nan_counts[nan_counts > threshold].index.tolist()

print(f"Columns to drop (more than 50% NaN values): {columns_to_drop}")
print(f"Number of columns to drop: {len(columns_to_drop)}")
```

```python
df_combined_cleaned = df_combined.drop(columns=columns_to_drop)

print(f"Shape of df_combined after dropping columns: {df_combined_cleaned.shape}")
print(f"Number of remaining columns: {df_combined_cleaned.shape[1]}")
```

```python
nan_counts = df_combined_cleaned.isnull().sum()
nan_columns = nan_counts[nan_counts > 0]

print("Columns with NaN values and their counts in df_combined_cleaned:")
display(nan_columns)
```

```python
nan_threshold_20000 = 20000
columns_to_drop_20k = nan_columns[nan_columns > nan_threshold_20000].index.tolist()

print(f"Columns to drop (more than 20,000 NaN values): {columns_to_drop_20k}")
print(f"Number of columns to drop: {len(columns_to_drop_20k)}")
```

```python
df_final_cleaned = df_combined_cleaned.drop(columns=columns_to_drop_20k)

print(f"Shape of the final cleaned DataFrame: {df_final_cleaned.shape}")
print(f"Number of remaining columns: {df_final_cleaned.shape[1]}")
```

```python
nan_counts = df_final_cleaned.isnull().sum()
nan_columns = nan_counts[nan_counts > 0]

print("Columns with NaN values and their counts in df_final_cleaned:")
display(nan_columns)
```

```python
columns_to_check_for_nan_rows = ['교육재학코드', '경제활동상태코드', '가구원소득구간코드']

# Drop rows where any of the specified columns have NaN values
df_rows_cleaned = df_final_cleaned.dropna(subset=columns_to_check_for_nan_rows)

print(f"Shape of the DataFrame after dropping rows with NaN values: {df_rows_cleaned.shape}")
print("Head of the DataFrame after dropping rows:")
display(df_rows_cleaned.head())
```

```python
list(df_rows_cleaned.columns)
```

```python
day_of_week_counts = df_rows_cleaned['요일구분코드'].value_counts()
print("Count of each value in '요일구분코드' column:")
display(day_of_week_counts)
```

```python
columns_to_drop_specific = [
    '(주행동시간대) 오전 00:00.1',
    '(주행동시간대) 오전 00:10.1',
    '(주행동시간대) 오전 00:20.1',
    '(주행동시간대) 오전 00:30.1',
    '(주행동시간대) 오전 00:40.1',
    '(주행동시간대) 오전 00:50.1'
]

df_rows_cleaned = df_rows_cleaned.drop(columns=columns_to_drop_specific, errors='ignore')

print(f"Shape of the DataFrame after dropping specific columns: {df_rows_cleaned.shape}")
print("Head of the DataFrame after dropping specific columns:")
display(df_rows_cleaned.head())
```

```python
marital_status_counts = df_rows_cleaned['혼인상태코드'].value_counts()
print("Count of each value in '혼인상태코드' column:")
display(marital_status_counts)
```

```python
df_filtered_marital_status = df_rows_cleaned[df_rows_cleaned['혼인상태코드'].isin([1, 2])]

print(f"Shape of the DataFrame after filtering by marital status: {df_filtered_marital_status.shape}")
print("Head of the DataFrame after filtering:")
display(df_filtered_marital_status.head())
```

```python
age_counts = df_filtered_marital_status['연령'].value_counts().sort_index()
print("Count of each value in '연령' column:")
display(age_counts)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create the bar plot
plt.figure(figsize=(15, 7))
sns.barplot(x=age_counts.index, y=age_counts.values, palette='viridis')
plt.title('Distribution of Age in Filtered Data')
plt.xlabel('Age')
plt.ylabel('Count')
plt.xticks(rotation=90) # Rotate x-axis labels if they are too crowded
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
age_code_counts = df_filtered_marital_status['연령코드'].value_counts().sort_index()
print("\nCount of each value in '연령코드' column:")
display(age_code_counts)
```

```python
# Filter to keep only rows where '연령코드' is 3 or 4
df_age_filtered = df_filtered_marital_status[df_filtered_marital_status['연령코드'].isin([3, 4])]

print(f"Shape of the DataFrame after filtering to keep 연령코드 3 and 4: {df_age_filtered.shape}")
print("Head of the DataFrame after filtering:")
display(df_age_filtered.head())

# Recalculate age_counts and age_code_counts with the new filtered DataFrame
age_counts = df_age_filtered['연령'].value_counts().sort_index()
age_code_counts = df_age_filtered['연령코드'].value_counts().sort_index()

print("\nUpdated Count of each value in '연령' column:")
display(age_counts)

print("\nUpdated Count of each value in '연령코드' column:")
display(age_code_counts)
```
