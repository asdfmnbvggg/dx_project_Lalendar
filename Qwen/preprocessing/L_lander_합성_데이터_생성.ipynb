```python
import json
import time
import pandas as pd
from tqdm import tqdm
from concurrent.futures import ThreadPoolExecutor, as_completed

# ==========================================
# ⚙️ [기획자용 컨트롤러 설정]
# ==========================================
IS_TEST = False     # 🟢 True: 1번만 호출(10개 테스트) / False: 5000건 전체 생성 가동!
MAX_WORKERS = 5     # 동시에 보낼 API 요청 수 (안정적인 속도 유지)
CHUNK_SIZE = 10     # 한 번에 묶어서 보낼 페르소나 개수 (RPD 절약의 핵심)
# ==========================================

# 1. 고도화된 묶음 처리용 LLM 프롬프트 템플릿
def generate_detailed_schedule_prompt(persona_info_json_list):
    prompt = f"""
너는 가상의 사용자 프로필을 다각도로 분석해 맞춤형 캘린더 일정과 그날의 날씨 환경 데이터를 생성하는 고도화된 데이터 엔지니어다.
아래 제공된 [사용자 프로필 배열]에는 여러 명의 페르소나 데이터(딕셔너리)가 담겨 있다.

[사용자 프로필 배열 (JSON List)]
{persona_info_json_list}

[데이터 생성 시 반드시 반영할 핵심 지침]
- 일정명(event_title)은 보고서 목차나 AI가 요약한 태그 같아선 절대 안 되며, 30~40대 한국 직장인/기혼자가 스마트폰 캘린더 앱에 바쁘게 타이핑할 법한 '매우 현실적이고 파편화된 단어'여야 함.
- 집에서 수행되는 일상적인 행동(예: 홈카페, 독서, 휴식)은 캘린더에 적지 않으므로 생성하지 않음. 단, 집들이, 양가 부모님 방문 등 특별한 이벤트는 가능.
- 30~49세 기혼 및 배우자 동거 가구 특성을 반영하여, 개인 일정 외에도 배우자나 자녀(자녀가 있다고 가정할 시)가 엮인 일정을 포함할 것.

[일정명(event_title) 작성 패턴 가이드 및 예시]
- ❌ 나쁜 예시 (AI스러운 억지 요약): 홈 카페에서 커피, 주말이벤트, 주말 가족 캠핑, 실내 대청소
- 🟢 좋은 예시 (장소/브랜드 + 목적): 이마트 장보기, 스타필드(애들옷), 강릉 1박2일, 양양 서핑, 주말 가족 캠핑
- 🟢 좋은 예시 (대상 + 행사): 첫째 운동회, 장인어른 생신, 동기 모임, 부서 회식, 와이프 결기(결혼기념일)
- 🟢 좋은 예시 (직관적인 활동): 세차, 골프 라운딩, 풋살, 건강검진

[데이터 생성 및 제약 조건]
1. temp_id: 입력받은 페르소나의 temp_id를 매칭 오류 방지를 위해 결과 객체에 그대로 유지할 것.
2. event_title (일정명): 프로필 항목이 반영된 구체적이고 간결한 행위 명사형
3. event_date (일정 날짜): 2026년 이내의 가상 날짜를 'YYYY-MM-DD' 형식으로 생성
4. event_start_time / event_end_time (시작/종료 시간): 'HH:MM' 24시간 형식 (종료 시간은 시작보다 미래)
5. day_temp (온도): 계절에 걸맞은 섭씨 온도 수치 (숫자만, 예: 23.8)
6. day_humidity (습도): 상대 습도 백분율 수치 (0~100 사이 정수, 예: 65)
7. day_dust (미세먼지): PM10 미세먼지 농도 수치 (숫자만, 예: 25)

출력 형식은 반드시 아래 제시된 구조의 JSON 포맷이어야만 해. 설명 없이 오직 JSON 배열만 반환해라.

[출력 JSON 포맷 예시]
{{
  "results": [
    {{
      "temp_id": 0,
      "event_title": "주말 가족 캠핑",
      "event_date": "2026-05-16",
      "event_start_time": "14:00",
      "event_end_time": "17:00",
      "day_temp": 24.5,
      "day_humidity": 45,
      "day_dust": 28
    }}
  ]
}}
"""
    return prompt

# 2. 데이터 전처리 및 청크 분할 구역
selected_columns = [
    'professional_persona', 'travel_persona', 'culinary_persona', 'family_persona',
    'persona', 'hobbies_and_interests_list', 'sex', 'age', 'marital_status',
    'family_type', 'housing_type'
]

final_df = final_df.copy()
final_df['temp_id'] = range(len(final_df))

# 10개씩 페르소나 데이터 묶기
chunks = [final_df[i:i + CHUNK_SIZE] for i in range(0, len(final_df), CHUNK_SIZE)]

# 🔥 [스위치 로직] 테스트 모드일 경우 맨 앞의 1묶음(10개)만 남김
if IS_TEST:
    chunks = chunks[:1]
    print(f"🛠️ [테스트 모드 활성화] 딱 1번만 API를 호출하여 10개의 데이터를 합성합니다.")
else:
    print(f"🚀 [전체 생성 모드 가동] 총 {len(chunks)}번의 API 호출로 {len(final_df)}개의 데이터 합성을 시작합니다.")

generated_schedules = []

# 3. 묶음 단위 쓰레드 실행 함수 정의
def generate_schedules_for_chunk(chunk, max_retries=3):
    persona_list = []
    for idx, row in chunk.iterrows():
        persona_dict = {"temp_id": int(row['temp_id'])}
        for col in selected_columns:
            value = row.get(col)
            persona_dict[col] = value if pd.notna(value) else None
        persona_list.append(persona_dict)

    persona_json_str = json.dumps(persona_list, ensure_ascii=False, indent=2)

    for attempt in range(max_retries):
        try:
            response = client.chat.completions.create(
                model="gpt-4o-mini",
                messages=[{"role": "user", "content": generate_detailed_schedule_prompt(persona_json_str)}],
                response_format={"type": "json_object"},
                temperature=0.8
            )

            result_json = json.loads(response.choices[0].message.content)

            # ⚠️ 프롬프트 지침에 맞게 "results" 키로 받아오도록 정합성 수정
            schedules = result_json.get("results", [])

            # 받아온 결과에 원본 행 번호(Index) 복원 및 매칭 안정화
            for schedule in schedules:
                t_id = schedule.get("temp_id")
                if t_id is not None:
                    orig_idx = chunk[chunk['temp_id'] == t_id].index
                    if not orig_idx.empty:
                        schedule["original_persona_index"] = orig_idx[0]

            return schedules

        except Exception as e:
            if attempt < max_retries - 1:
                time.sleep(2 ** attempt)
            else:
                print(f"⚠️ 묶음 생성 실패 (temp_id {persona_list[0]['temp_id']} 시작 구간): {e}")
                return []

# 4. 병렬 처리 오케스트레이션 가동
with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
    futures = {executor.submit(generate_schedules_for_chunk, chunk): i for i, chunk in enumerate(chunks)}

    for future in tqdm(as_completed(futures), total=len(futures)):
        generated_schedules.extend(future.result())

# 5. 데이터 병합 및 최종 저장 구역
df_generated = pd.DataFrame(generated_schedules)

if not df_generated.empty:
    # 꼬임 방지를 위해 temp_id 기준으로 원본 페르소나와 1:1 병합
    x_df = pd.merge(final_df, df_generated, on='temp_id', how='inner')
    x_df = x_df.drop(columns=['temp_id']) # 분석에 불필요한 임시 ID 제거

    column_order = [
        'original_persona_index', 'event_title', 'event_date',
        'event_start_time', 'event_end_time', 'day_temp', 'day_humidity', 'day_dust'
    ]
    all_columns = column_order + [c for c in x_df.columns if c not in column_order]
    x_df = x_df[all_columns]

    print(f"\n🎉 데이터셋 생성 완료! (성공 건수: {len(x_df)}건 / 요청 건수: {len(chunks)*CHUNK_SIZE}건)")

    output_file = "test_X_dataset.csv" if IS_TEST else "calendar_weather_X_dataset_final.csv"
    x_df.to_csv(output_file, index=False, encoding='utf-8-sig')
    print(f"💾 '{output_file}' 파일로 저장 완료!")
else:
    print("❌ AI가 결과를 반환하지 못해 데이터프레임 생성이 무산되었습니다. 프롬프트나 API Key를 확인하세요.")

x_df.head(10)
```

```python
import json
import pandas as pd
from tqdm import tqdm
from openai import OpenAI

# 0. x_df1 데이터에 겹치지 않는 고유 ID 부여 (중복 병합 완벽 차단)
if 'row_id' not in x_df1.columns:
    x_df1['row_id'] = range(len(x_df1))

# 1. 프롬프트 정의
def generate_batch_y_value_prompt(x_data_list_str):
    prompt = f"""
너는 스마트홈 가전 제어 AI를 학습시키기 위한 최고 수준의 데이터 엔지니어다.
아래는 여러 명의 [입력 데이터 (X값)]가 담긴 JSON 배열이다.
각 데이터의 상황(일정, 날씨)을 깊이 있게 분석하여, 해당 일정에 가장 상식적이고 최적화된 가사일과 가전 제어 명령(Y값)을 추론해라.

[입력 데이터 배열 (X값)]
{x_data_list_str}

[Rule 1: Mandatory Trigger (강제 할당)]
- IF 일정 IN ["운동", "걷기·산책", "낚시·사냥"]:
  THEN 직후 스케줄에 "세탁기" 및 "건조기" 필수 포함
- IF 일정 IN ["식사하기"]:
  THEN 직후 스케줄에 "식기세척기" 필수 포함

[Rule 2: Optional Trigger (조건부 할당)]
- IF 일정 종료 IN ["개인위생 및 외모 관리", "출근 및 업무 활동", "종교/의료 활동"]:
  THEN 직후 스케줄에 "로봇청소기" 또는 "공기청정기" 선택적 포함

[★ 엄격한 제한 사항: 가전 및 작동 모드]
AI가 작동시킬 가전과 모드는 반드시 아래 제공된 6종의 가전(영문)과 작동 모드(한글) 안에서만 선택해야 한다.
- 세탁기 (washer) : 표준, 울, 이불삶음, 아기옷, 조용조용, 스피드, 기능성, 통살균, 헹굼+탈수
- 건조기 (dryer) : 표준, 타월, 셔츠, 울, 이불, 소량급속, 기능성, 선반건조, 패딩케어
- 에어컨 (air_conditioner) : 냉방, 파워, 제습, 송풍, 취침
- 식기세척기 (dishwasher) : 표준, 강력, 급속, 섬세, 살균, 통살균
- 공기청정기 (air_purifier) : 자동, 터보, 취침, 에코, 펫, 순환
- 로봇청소기 (robot_cleaner) : 자동, 구역, 스팟, 물걸레

[★ 일정 유형별 최적 가전 매칭 가이드 (반드시 준수)]
단순히 특정 가전을 무작위로 고르지 말고, 일정의 성격에 맞춰 1순위로 필요한 가전을 논리적으로 매칭해라.
1. 야외 활발 활동 및 운동 (예: 탐방, 서핑, 등산, 산책, 운동회 등)
   - 귀가 후 땀/흙/소금기 오염 해결 -> [사후] 세탁기(washer) 기능성/강력 모드
   - 야외 활동 후 지친 상태로 귀가 시 쾌적함 제공 -> [사전 준비] 에어컨(air_conditioner) 냉방/제습 또는 공기청정기(air_purifier)
   - 외출로 집이 비어있는 시간 활용 -> [외출 중] 로봇청소기(robot_cleaner) (작동 시간을 외출 시간 내로 설정)
2. 손님 맞이 및 식사/파티 (예: 부모님 방문, 지인 초대, 홈파티)
   - 손님 도착 전 쾌적한 환경 조성 -> [사전 준비] 로봇청소기(robot_cleaner), 공기청정기(air_purifier)
   - 행사/식사 종료 후 대량의 식기 정리 -> [사후] 식기세척기(dishwasher) 강력/살균
3. 일반 외출/휴식 (예: 장보기, 영화 관람, 외식)
   - 외출복 먼지/냄새 케어 -> [사후] 세탁기(washer) 또는 건조기(dryer) 패딩케어/셔츠
   - 더운/습한 날씨 귀가 대비 -> [사전 준비] 에어컨(air_conditioner) 제습/파워

[출력 데이터 작성 지침]
- JSON 배열의 Key 이름은 "results" 로 할 것.
- 각 객체는 다음 9개의 필수 Key를 가져야 함:
  1. "event_reasoning": 위 매칭 가이드를 참고하여 일정의 성격을 분석하고, 왜 이 타이밍(사전/사후/외출중)에 가전 제어가 필요한지 구체적인 이유를 서술해라. (예시를 그대로 베끼지 말고 상황에 맞게 창작할 것)
  2. "expected_task_reasoning": 위 상황을 해결하기 위한 구체적인 조치를 적어라.
  3. "weather_reasoning": 날씨(온도/습도/미세먼지) 데이터를 근거로 활용해라. (온도가 높으면 에어컨 근거로, 습도가 높으면 제습/건조기 근거로 활용)
  4. "task_name": 캘린더에 일정으로 등록될 수 있는 자연스럽고 구체적인 작업명.
  5. "task_appliance": 일정 맥락에 가장 최적화된 6종 가전 중 택 1.
  6. "task_appliance_mode": 상황과 가전에 맞는 최적의 특수 모드 택 1.
  7. "task_date": 작동 날짜 ('YYYY-MM-DD')
  8. "task_start_time": 작동 시작 시간 ('HH:MM', ★논리적으로 완벽해야 함. 사전 준비는 일정 전, 사후는 일정 후, 외출 중 청소는 일정 시간 사이로 배치할 것)
  9. "task_end_time": 작동 종료 시간 ('HH:MM')
"""
    return prompt

# 2. 데이터 쪼개기 및 전체 생성 설정 (배치 사이즈 20 적용)
CHUNK_SIZE = 20
y_generated_results = []

# 전체 x_df1 데이터를 지정된 청크로 분할합니다.
chunks = [x_df1[i:i + CHUNK_SIZE] for i in range(0, len(x_df1), CHUNK_SIZE)]

print(f"🚀 [전체 생성 모드] 총 7,000건의 데이터를 {len(chunks)}번의 API 호출로 안전하게 합성합니다.")

# 지정된 7개 칼럼 목록
target_columns = ['event_title', 'event_date', 'event_start_time', 'event_end_time', 'day_temp', 'day_humidity', 'day_dust']

# 3. 묶음 단위로 API 호출
for chunk_idx, chunk in enumerate(tqdm(chunks)):

    # AI에게 전송할 때는 지정된 7개 칼럼만 전송하여 토큰 낭비 및 노이즈 차단
    chunk_filtered = chunk[target_columns]

    chunk_list = chunk_filtered.to_dict(orient='records')
    x_data_list_str = json.dumps(chunk_list, ensure_ascii=False, indent=2)

    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": generate_batch_y_value_prompt(x_data_list_str)}],
            response_format={ "type": "json_object" },
            temperature=0.7
        )

        result_json = json.loads(response.choices[0].message.content)
        results_array = result_json.get("results", [])

        # 파이썬이 원본 번호표(row_id)를 순서대로 다시 매칭
        for i, result in enumerate(results_array):
            if i < len(chunk):
                result['row_id'] = chunk.iloc[i]['row_id']
                y_generated_results.append(result)

    except Exception as e:
        print(f"⚠️ {chunk_idx}번째 묶음 에러 발생: {e}")
        continue

# 4. 결과 변환 및 결합
y_df = pd.DataFrame(y_generated_results)

# 병합 기준을 row_id로 설정하여 원본 x_df1과 1:1 완벽 매칭 보장
final_master_df = pd.merge(x_df1, y_df, on='row_id', how='inner')

# 임시로 생성했던 고유 식별자 row_id 삭제
final_master_df = final_master_df.drop(columns=['row_id'])

print(f"\n🎉 [전체 생성 완료] 오류 없이 1:1로 매칭된 {len(final_master_df)}건의 최종 마스터 데이터셋 완성!")

# 파일명 지정 및 저장
final_master_df.to_csv("/content/drive/MyDrive/final_master_df2.csv", index=False, encoding='utf-8-sig')

# 결과 출력
final_master_df.head()
```
